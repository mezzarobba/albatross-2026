## To fill

```julia
R,(c1,c2,c3,s1,s2,s3) = polynomial_ring(QQ, [:c1,:c2,:c3,:s1,:s2,:s3])
```

```julia
K = 1//2 * [ c1*c2*(3*c3+4) - s1*(3*s3+2) + c1, s1*c2*(3*c3+4) + c1*(3*s3+2) + s1, -s2*(3*c3+4) ]
V = Ideal([ c1^2 + s1^2 - 1, c2^2 + s2^2 - 1, c3^2 + s3^2 - 1 ])
K, V
```

```julia
critK[end] = collect(fact)[end][1]
```

```julia
dimension(Ideal(critK)), hilbert_degree(Ideal(critK))
```

## How to simplify to avoid luckyness

```julia
# Simplification
R1,(c2,c3,s2,s3) = polynomial_ring(QQ, [:c2,:c3,:s2,:s3])
K1, V1, critK1 = [ [ f(1,c2,c3,0,s2,s3) for f in E if !iszero(f(1,c2,c3,0,s2,s3))] for E in [K, V.gens, critK] ]
```

```julia
dimension(Ideal(V1))
```

```julia
dimension(Ideal(critK1)), hilbert_degree(Ideal(critK1))
```

```julia
real_solutions(Ideal(vcat(V1, [K1[1]-457//584, K1[2]])))
```

```julia
R2,(c2,c3,s2,s3, rho, z) = polynomial_ring(QQ, [:c2,:c3,:s2,:s3,:rho,:z])
K2, V2, critK2 = [ [ f(c2,c3,s2,s3) for f in E if !iszero(f(c2,c3,s2,s3))] for E in [K1, V1, critK1] ]
KcritK2 = Ideal(vcat([z - K2[end], rho^2 - K2[1]^2 - K2[2]^2], critK2))
```

```julia
G1 = eliminate(KcritK2, 4)
```

```julia
plop, (rho,z) = polynomial_ring(QQ, [:rho,:z])
svalK1 = [ g(1,1,1,1,rho,z) for g in G1 ]
```

```julia
Wsval1 = points_per_components(QQMPolyRingElem[], [rho], svalK1, info_level=5)
```

Idée : on pourrait déjà ne prendre qu'un point par composante connexe ici car on travaille avec seulement deux variables !
On peut donc le faire avec des outils simples (théorème des intersection) ou encore graphiquement

```julia
R_rab,(u,c2,c3,s2,s3) = polynomial_ring(QQ, [:u, :c2,:c3,:s2,:s3])
critK1_rab = [ f(c2,c3,s2,s3) for f in critK1 ]
Vmcrit1_rab = vcat(critK1_rab[1:end-1], 1 - u*critK1_rab[end])
K1_rab = [ f(c2,c3,s2,s3) for f in K1 ]
```

```julia
function plot_graph(G)
  P = build_graph_data(G)
  plot(legend=false)
  E = P.edge_group
  plot!(E.edges, color=E.color, width = E.width)
  [scatter!(V.vertices, color=V.color, marker = V.marker) for V in P.point_groups]
  plot!()
  #gui()
end

function plot_graphs(CG)
  CP = build_graph_data(CG, vert=false)
  plot(legend=false)
  for P in CP
      E = P.edge_group
      plot!(E.edges, color=E.color, width = E.width)
      [ scatter!(V.vertices, color=V.color, marker = V.marker) for V in P.point_groups ]
  end
  plot!()
  #gui()
end
```
